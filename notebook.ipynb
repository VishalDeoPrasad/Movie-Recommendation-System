```python
df = df[['id', 'title', 'genre', 'overview']]
df
```

```python
df['movie_desc'] = df.overview + df.genre
```

```python
df.head()
```

```python
df.drop(columns=['genre', 'overview'], inplace=True)
```

```python
df.head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=10000, stop_words='english')
cv
```

```python
vector = cv.fit_transform(df.movie_desc.values.astype('U'))
vector = vector.toarray()
```

```python
vector
```

```python
df.shape
```

```python
df.title
```

```python
from sklearn.metrics.pairwise import cosine_similarity
movie_similarity = cosine_similarity(vector)
```

```python
movie_similarity
```

```python
movie_similarity.shape
```

```python
movie_similarity[0]
```

```python
df.head(3)
```

```python
def recommendation(movie):
    index = df[df.title == movie].index[0]
    dist = sorted(list(enumerate(movie_similarity[index])), reverse=True, key=lambda x: x[1])
    for i in dist[:10]:
        print(df.iloc[i[0]].title)
```

```python
recommendation("Iron Man")
```

```python
recommendation("Star Wars")
```

```python
recommendation("Avengers: Age of Ultron")
```
